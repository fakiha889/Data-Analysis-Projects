# **Task 1 : Student dataset**
# **Description**
**The main purpose of this task is to understand student performance by using data.**
**I wanted to find out which students received Grade A, how many students had attendance above 80%, how class participation was related to grades, and whether submitting assignments on time was related to better grades.**

```python

import pandas as pd
import matplotlib.pyplot as plt
students = pd.DataFrame({
    "Student_ID": ["S001", "S002", "S003", "S004", "S005", "S006",
                   "S007", "S008", "S009", "S010", "S011", "S012"],

    "Attendance": [95, 88, 72, 60, 85, 92,
                   78, 55, 98, 81, 67, 90],

    "On_Time": ["Yes", "Yes", "No", "No", "Yes", "Yes",
                "No", "No", "Yes", "Yes", "No", "Yes"],

    "Participation": ["High", "Medium", "Low", "Low", "Medium", "High",
                      "Medium", "Low", "High", "Medium", "Low", "High"],

    "Grade": ["A", "B", "C", "Fail", "B", "A",
              "C", "Fail", "A", "B", "C", "A"]
})

print(students)
print(students.shape)
print(students.columns.tolist())
print(students.dtypes)
# Question 1: which students received Grade A?
b = students[students["Grade"]=="A"]["Student_ID"]
print(b)
# Question 2: how many students have attendance above 80%?
c = len(students[students["Attendance"]>80])
print(c)
# Question 3: How many students have High participation?
d = len(students[students["Participation"]=="High"])
print(d)
# Question 7: How many students submitted assignments late?
f = len(students[students["On_Time"]=="No"])
print(f)
# Question 9: What is the average attendance of students
# who received Grade A?
g = students[students["Grade"]=="A"]["Attendance"].mean()
print(g)
# # Question 10: What is the average attendance of students
# who failed?
h = students[students["Grade"]=="Fail"]["Attendance"].mean()
print(h)
# # Question 11: Among students with High participation,
# how many received Grade A?
k = len(students[(students["Participation"] == "High") &(students["Grade"] == "A")])
print(k)
# Question 12: Among students who submitted assignments on time,
# what grade is most common?
grade = students[students["On_Time"]=="Yes"]
common = grade["Grade"].mode()[0]
print(common)
#Is submitting assignments on time (On_Time)
# related to getting a higher grade?
print(pd.crosstab(students["On_Time"],students["Grade"]))

##### paticipation
level = ["High","Low","Medium"]
student_pct = {}
for i in level:
  student_data = students[students["Participation"]==i]
  pct = (student_data["Grade"]=="A").mean()*100
  student_pct[i] = pct
  print(f"{i}:{pct}  % of students with high class participation received Grade A")

  #### assignment
assignment = ["Yes","No"]
student_pct = {}
for i in assignment:
  data = students[students["On_Time"]==i]
  pct = (data["Grade"]=="A").mean()*100
  student_pct[i] = pct
  print(f"{i}:{pct}  % of students with assignment on time received Grade A")


  ###### attendance
high_attendance = students[students["Attendance"] >= 80]
pct = (high_attendance["Grade"] == "A").mean() * 100
print(f"{pct:.0f}  % of students with attendance >= 80 received Grade A")


#### matplotlib
plt.figure(figsize=(6, 3.6))
plt.bar(list(student_pct.keys()), list(student_pct.values()))
plt.title("% of students receiving Grade A")
plt.ylabel("% Grade A")
plt.ylim(0, 110)
plt.tight_layout()
plt.show()
```

# **Task 2 : Bank Dataset**

# **Description**

**I wanted to find out how many customers paid their bills online, how many customers used digital loan services, how customers used ATMs, and how many customers adopted the banking app.**

```python
import pandas as pd
import matplotlib.pyplot as plt

customers = pd.DataFrame({
    "Customer_ID": ["C001", "C002", "C003", "C004", "C005", "C006",
                    "C007", "C008", "C009", "C010", "C011", "C012"],

    "Pays_Bills_Online": ["Yes", "Yes", "No", "Yes", "No", "Yes",
                          "Yes", "No", "Yes", "No", "Yes", "No"],

    "Loan_Digital": ["Yes", "No", "No", "Yes", "No", "Yes",
                     "Yes", "No", "No", "No", "Yes", "No"],

    "ATM_Only": ["No", "Yes", "Yes", "No", "Yes", "No",
                 "No", "Yes", "No", "Yes", "No", "Yes"],

    "Adopts_App": ["Yes", "Yes", "No", "Yes", "No", "Yes",
                   "Yes", "No", "Yes", "No", "Yes", "No"]
})

print(customers)
print(customers.shape)
print(customers.columns.tolist())
# Question 1:
# How many customers adopted the mobile app?
a = len(customers[customers["Adopts_App"]=="Yes"])
print(a)
# Question 2:
# How many customers rely only on ATM services?
d = len(customers[customers["ATM_Only"]=="Yes"])
print(d)
# Question 5:
# Is paying bills online related to adopting the mobile app?
print(pd.crosstab(customers["Pays_Bills_Online"],customers["Adopts_App"]))
# Question 6:
# Is making digital loan payments related to adopting the mobile app?
print(pd.crosstab(customers["Loan_Digital"],customers["Adopts_App"]))
# Question 7:
# What percentage of customers who pay bills online adopted the mobile app?
h = customers[customers["Pays_Bills_Online"]=="Yes"]
n = (h["Adopts_App"]=="Yes").mean()*100
print(n)
# Question 9:
# What percentage of customers who make digital loan payments adopted the mobile app?
i = customers[customers["Loan_Digital"]=="Yes"]
j = (i["Adopts_App"]=="Yes").mean()*100
print(j)
# Question 10:
# Among customers who adopted the mobile app,
# how many also pay bills online?
g = len(customers[customers["Pays_Bills_Online"]=="Yes"])& (customers["Adopts_App"]=="Yes")
print(g)
information = ["Pays_Bills_Online","Loan_Digital","ATM_Only"]
customer_pct = {}
for  i in information:
  data = customers[customers[i]=="Yes"]
  pct = (data["Adopts_App"]=="Yes").mean()*100
  customer_pct[i] = pct
  print(f"{i} : {pct}% customer  Adopt app ")
## matplotlib
plt.figure(figsize=(6,3.6))
plt.bar(list(customer_pct.keys()),list(customer_pct.values()),color="#1F4E79")
plt.title("% of customer with each pay who were adopts mobile app")
plt.ylabel("% adopts mobile app")
plt.ylim(0,110)
plt.xticks(rotation = 15)
plt.tight_layout()
plt.show()
```

# **Task 3 :Grocery dataset**
# **Description**
**I wanted to find out how many customers purchased bread, butter, jam, milk, and eggs, which grocery items were purchased most often, and which items were commonly purchased together.**

```python

import pandas as pd
import matplotlib.pyplot as plt

orders = pd.DataFrame({
    "Order_ID": ["O001", "O002", "O003", "O004", "O005", "O006",
                 "O007", "O008", "O009", "O010", "O011", "O012"],

    "Bread":  ["Yes", "Yes", "No",  "Yes", "No",  "Yes",
               "Yes", "No",  "Yes", "No",  "Yes", "Yes"],

    "Butter": ["Yes", "No",  "No",  "Yes", "Yes", "No",
               "Yes", "No",  "Yes", "No",  "No",  "Yes"],

    "Jam":    ["No",  "Yes", "No",  "Yes", "No",  "No",
               "Yes", "Yes", "No",  "No",  "Yes", "Yes"],

    "Milk":   ["Yes", "Yes", "Yes", "No",  "Yes", "Yes",
               "No",  "Yes", "Yes", "No",  "Yes", "Yes"],

    "Eggs":   ["No",  "Yes", "Yes", "Yes", "No",  "Yes",
               "No",  "Yes", "Yes", "Yes", "No",  "Yes"]
})

print(orders)
print(orders.shape)
print(orders.columns.tolist())
# Question 1:
# How many orders contain Bread?
a = len(orders[orders["Bread"]=="Yes"])
print(a)
# Question 2:
# which customers orders contain Butter?
b = orders[orders["Butter"]=="Yes"]["Order_ID"]
print(b)
# Question 3:
# Is buying Bread related to buying Butter?
print(pd.crosstab(orders["Bread"],orders["Butter"]))
# Question 5:
# Is buying Bread related to buying Jam?
print(pd.crosstab(orders["Bread"],orders["Jam"]))
# Question 6:
# Is buying Milk related to buying Eggs?
print(pd.crosstab(orders["Milk"],orders["Eggs"]))
# Question 7:
# What percentage of orders containing Bread also contain Butter?
d = orders[orders["Bread"]=="Yes"]
e = (d["Butter"]=="Yes").mean()*100
print(e)
# Question 8:
# What percentage of orders containing Bread also contain Jam?
f = orders[orders["Bread"]=="Yes"]
g = (d["Jam"]=="Yes").mean()*100
print(g)


order1 = ["Bread"]
order2 = ["Butter"]
order3 = ["Eggs"]
order1_pct = {}
order2_pct = {}
order3_pct = {}

for i in order1:
  order_data = orders[orders["Bread"]=="Yes"]
  pct = (order_data["Jam"]=="Yes").mean()*100
  order1_pct[i] = pct
  print(f"{i} : {pct} people buy bread and jam")

for i in order2:
  order_data = orders[orders["Butter"]=="Yes"]
  pct = (order_data["Bread"]=="Yes").mean()*100
  order2_pct[i] = pct
  print(f"{i} : {pct} people buybutter and bread")

for i in order3:
  order_data = orders[orders["Eggs"]=="Yes"]
  pct = (order_data["Milk"]=="Yes").mean()*100
  order3_pct[i] = pct
  print(f"{i} : {pct} buy both")

### matplolib
plt.figure(figsize=(6,3.6))
plt.bar(list(order1_pct.keys()),list(order1_pct.values()),color="#1F4E79")
plt.title("% of customer buy bread and jan ")
plt.ylabel("% of people")
plt.ylim(0,110)
plt.xticks(rotation = 15)
plt.tight_layout()
plt.show()
plt.figure(figsize=(6,3.6))
plt.bar(list(order2_pct.keys()),list(order2_pct.values()),color="#1F4E79")
plt.title("% of customer buy bread and butter ")
plt.ylabel("% percentage of people")
plt.ylim(0,110)
plt.xticks(rotation = 15)
plt.tight_layout()
plt.show()
plt.figure(figsize=(6,3.6))
plt.bar(list(order3_pct.keys()),list(order3_pct.values()),color="#1F4E79")
plt.title("% of customer buy milk and egg ")
plt.ylabel("% of people")
plt.ylim(0,110)
plt.xticks(rotation = 15)
plt.tight_layout()
plt.show()




```

# **Task 4 :Hospital dataset**
# **Description**
 **In this task,we will indentify which patients recommend x_ray**

```python

import pandas as pd
import matplotlib.pyplot as plt

patients = pd.DataFrame({
     "Patient_ID": ["P001", "P002", "P003", "P004", "P005", "P006", "P007", "P008"],
     "Cough": ["Yes", "Yes", "No",  "Yes", "No",  "Yes", "No",  "Yes"],
     "Fever":["Yes", "No",  "No",  "Yes", "Yes", "Yes", "No",  "No"],
     "Breathing_Difficulty": ["No", "Yes", "No", "Yes", "No", "No", "No", "Yes"],
     "Chest_Pain": ["No",  "No",  "No",  "Yes", "No",  "No",  "Yes", "Yes"],
     "Xray_Recommended": ["Yes", "Yes", "No", "Yes", "No", "Yes", "No", "Yes"], })
print(patients)
print(patients.shape)
print(patients.columns.tolist())
# Step 1: Ask and observe. How many patients got an X-ray, and is cough related to it?
# Question 1: How many patients were recommended an X-ray?
print(patients["Xray_Recommended"].value_counts())
# Question 2: Is having a cough related to being recommended an X-ray?
print(pd.crosstab(patients["Cough"], patients["Xray_Recommended"]))
#Compare every symptom.
symptoms = ["Cough", "Fever", "Breathing_Difficulty", "Chest_Pain"]
symptom_pct = {} # empty dictionary to store the results
for symptom in symptoms:
 with_symptom = patients[patients[symptom] == "Yes"] # keep rows with this symptom
 pct = (with_symptom["Xray_Recommended"] == "Yes").mean() * 100
 symptom_pct[symptom] = pct
 print(f"{symptom}: {pct:.0f}% of patients with this symptom were sent for X-ray")
 # Visualise and state the insight
plt.figure(figsize=(6, 3.6))
plt.bar(list(symptom_pct.keys()), list(symptom_pct.values()), color="#1F4E79")
plt.title("% of patients with each symptom who were sent for X-ray")
plt.ylabel("% sent for X-ray")
plt.ylim(0, 110)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```
